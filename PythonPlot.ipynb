## 簡単な使い方

`matplotlib` の名前の由来となった MATLAB のプロットと同じような使い方

### plot

曲線のグラフ，関数のグラフ

```python
import numpy as np
from numpy import pi, sin, cos
import matplotlib.pyplot as plt

xs = np.linspace(0, 2 * pi)
plt.plot(xs, sin(xs))
```

何をやっているか覗いてみる。

```python
xs
```

`numpy` の数値関数は，
`numpy` の配列 (`ndarray`) が与えられると成分ごとに計算した結果を返す。

```python
sin(xs)
```

`plt.plot` が描いているのは，両者が作る座標点を滑らかに結んだものである。

matplotlib のコーディング例を見ると
X座標列と Y座標列を，それぞれ変数 x, y で表しているものが多いが，
それだと 1個の座標値 (x, y) と誤解してしまう。

この文書では，データ列だということを明確にするために xs, ys という変数名を使う。
x-coordination sequence, y-coordination sequence の略だが，
x 座標の複数形，y 座標の複数形と思って読めばよい。

```python
import numpy as np
from numpy import pi, sin, cos
import matplotlib.pyplot as plt

xs = np.linspace(0, 2 * pi)

plt.plot(xs, sin(xs))
plt.plot(xs, cos(xs))
# 最後に plt.show() が暗黙的に呼び出される
```

x座標，y座標のペアを複数並べることができる。

```python
plt.plot(xs, sin(xs), xs, cos(xs))
```

グラフを1つ描いては `plt.show()` を実行し，別のグラフを描いてはまた `plt.show()` を実行すると，複数のグラフが描かれる。

```python
xs = np.linspace(0, 2 * pi)

# １つ目のグラフを描く
plt.plot(xs, sin(xs))
plt.show() # ここで１つ目のグラフ出力
  
# ２つ目のグラフを描く
plt.plot(xs, cos(xs))
plt.show() # ここで２つ目のグラフ出力
```

## 領域の塗りつぶし

`fill_between` を使うと塗り潰しができる。

```python
import numpy as np
from numpy import pi, sin, cos
import matplotlib.pyplot as plt

xs = np.linspace(0, 2 * pi)
ys = sin(xs)
x_axis = np.zeros_like(xs)

plt.plot(xs, ys, color = "black")
plt.plot(xs, x_axis, color = "black")
plt.fill_between(xs, ys, x_axis)
```

塗りつぶす区間を指定する。

```python
import numpy as np
from numpy import pi, sin, cos
import matplotlib.pyplot as plt

ys = sin(xs)
x_axis = np.zeros_like(xs)

plt.plot(xs, ys, "black")
plt.plot(xs, x_axis, "black")
plt.fill_between(xs, ys, x_axis, where=(ys > 0.5))
```

ここで `where` 引数 `ys > 0.5` の部分は次と同じ

```python
np.array([y > 0.5 for y in ys])
```

`np.logical_and` や `np.logical_or` を使うことで，複雑な条件を記述できる。

```python
import numpy as np
from numpy import pi, sin, cos
import matplotlib.pyplot as plt

xs = np.linspace(0, 2 * pi)
ys = sin(xs)
x_axis = np.zeros_like(xs)

plt.plot(xs, ys, "black")
plt.plot(xs, x_axis, "black")
plt.fill_between(xs, ys, x_axis, where=np.logical_and(2 < xs, xs < 4))
```

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt

xs = np.linspace(-3, 3)
ys = norm.pdf(xs)
x_axis = np.zeros(len(xs))

plt.plot(xs, ys, color = "black", label = "normal dist")
plt.plot(xs, x_axis, "black")
k = 1.96 * norm.std()
plt.fill_between(xs, ys, x_axis,
                 where = np.logical_or(xs < -k, k < xs),
                 label = "significant level")
plt.legend()
```

軸の表示を変える。

枠線を描かず，メモリが書かれた軸は $x = 0$, $y = 0$ にする。

```python
import numpy as np
from numpy import pi, sin, cos
import matplotlib.pyplot as plt

xs = np.linspace(0, 2 * pi)

plt.yticks([-1, -.5, .5, 1.0])
plt.plot(xs, sin(xs))
plt.plot(xs, cos(xs))

spines = plt.gca().spines
spines['left'].set_position('zero')
spines['bottom'].set_position('zero')
spines['right'].set_visible(False)
spines['top'].set_visible(False)
```

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt

xs = np.linspace(-4, 4)
ys = norm.pdf(xs)

plt.plot(xs, ys)
plt.yticks([0.1, 0.2, 0.3, 0.4])

spines = plt.gca().spines
spines['left'].set_position('zero')
spines['bottom'].set_position('zero')
spines['right'].set_visible(False)
spines['top'].set_visible(False)
```
